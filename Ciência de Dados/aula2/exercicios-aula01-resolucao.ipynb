## 2 · Aprovado, exame ou reprovado

Dada a lista `notas = [8.5, 4.0, 6.2, 9.1, 2.7]`, percorra com `for` e classifique cada
nota com `if/elif/else`: ≥ 7 aprovado · ≥ 5 exame · abaixo, reprovado. Use `enumerate`
para numerar a partir de 1.

```
Aluno 1: aprovado
Aluno 2: reprovado
Aluno 3: exame
...
```

````python
notas = [8.5, 4.0, 6.2, 9.1, 2.7]

for i, nota in enumerate(notas, start=1):
    if nota >= 7:
        situacao = "aprovado"
    elif nota >= 5:
        situacao = "exame"
    else:
        situacao = "reprovado"
    print(f"Aluno {i}: {situacao}")
````

## 3 · Fatiamento — sem nenhum laço

Dada `x = [12, 7, 45, 3, 28, 91, 16, 54, 8, 33]`, imprima usando **apenas fatiamento**
(nenhum `for`):

- os 3 primeiros elementos
- os 3 últimos
- os elementos de posição par (0, 2, 4…)
- a lista invertida
- tudo menos o primeiro e o último

````python
x = [12, 7, 45, 3, 28, 91, 16, 54, 8, 33]

# x[início:fim]        o fim é EXCLUSIVO — a posição do fim não entra
# x[início:fim:passo]  campo vazio vira o padrão: início 0, fim len(x), passo 1

print(x[:3])     # início vazio = 0      · fim 3, exclusivo   → posições 0, 1, 2
print(x[-3:])    # -3 = terceira de trás para frente · fim vazio = até o final
print(x[::2])    # início e fim vazios = tudo · passo 2 → posições 0, 2, 4, 6, 8
print(x[::-1])   # passo -1 = percorre de trás para frente → a lista invertida
print(x[1:-1])   # começa na posição 1 · termina antes da última (-1 exclusivo)
````

## 4 · Duas listas em paralelo

Com `nomes = ["Ana", "Bruno", "Carla"]` e `notas = [8.5, 6.0, 9.2]`, percorra as duas
juntas com `zip` e imprima `Ana: 8.5` (uma por linha). Depois, troque o conteúdo de duas
variáveis `a` e `b` **sem variável temporária**.

````python
nomes = ["Ana", "Bruno", "Carla"]
notas = [8.5, 6.0, 9.2]

for nome, nota in zip(nomes, notas):
    print(f"{nome}: {nota}")

a, b = 10, 20
a, b = b, a
print(a, b)
````

## 5 · Cadastro num dicionário

Crie `precos = {"café": 8.0, "pão": 0.75, "leite": 5.5}` e, em seguida:

- adicione `"queijo"` por 32.0
- aumente o preço do café para 9.0
- busque `"chocolate"` com `.get()` retornando 0 se não existir (sem `KeyError`!)
- percorra com `.items()` imprimindo `café — R$ 9.00`

````python
precos = {"café": 8.0, "pão": 0.75, "leite": 5.5}

precos["queijo"] = 32.0
precos["café"] = 9.0

print(precos.get("chocolate", 0))

for produto, preco in precos.items():
    print(f"{produto} — R$ {preco:.2f}")
````

## 7 · Do laço à comprehension

Reescreva o laço abaixo em **uma linha**, com list comprehension. Depois crie, com dict
comprehension, o dicionário `{n: n**2}` para os números de 1 a 5.

```python
pares = []
for n in range(1, 21):
    if n % 2 == 0:
        pares.append(n * 10)
```

````python
# [ expressão   for item in sequência   if condição ]
#       ↓              ↓                        ↓
#   o que guardar   de onde vem cada item    filtro: só entra quem passa
#                                            (o if é opcional)

pares = [n * 10 for n in range(1, 21) if n % 2 == 0]
print(pares)

# nas chaves e com "chave: valor" antes do for, o resultado é um dicionário
quadrados = {n: n ** 2 for n in range(1, 6)}
print(quadrados)
````

## 8 · Função com default

Escreva `media(notas, arredondar=False)` que retorna a média de uma lista — arredondada
para 1 casa (`round`) quando `arredondar=True`. Teste com `[8.5, 6.0, 9.3]`, chamando de
três formas: só posicional, com `arredondar=True`, e com **todos os argumentos nomeados**.

```
7.933333333333334
7.9
```

````python
def media(notas, arredondar=False):
    resultado = sum(notas) / len(notas)
    return round(resultado, 1) if arredondar else resultado


print(media([8.5, 6.0, 9.3]))
print(media([8.5, 6.0, 9.3], arredondar=True))
print(media(notas=[8.5, 6.0, 9.3], arredondar=True))
````

## 9 · Retorno múltiplo

Escreva `resumo(valores)` que retorna **três valores**: o menor, o maior e a média.
Aplique na lista `x` do exercício 3, desempacote o resultado em três variáveis e imprima
com f-string.

```
menor=3 · maior=91 · media=29.7
```

````python
x = [12, 7, 45, 3, 28, 91, 16, 54, 8, 33]   # a mesma lista do exercício 3


def resumo(valores):
    return min(valores), max(valores), sum(valores) / len(valores)


menor, maior, media_x = resumo(x)

print(f"menor={menor} · maior={maior} · media={media_x}")
````

## 10 · Caça às armadilhas

**Antes de rodar**, escreva num comentário o que cada trecho imprime. Depois rode,
confira, e explique os que errou. Por fim, **corrija o trecho C** usando o idioma correto.

```python
# A
a = [1, 2, 3]
b = a
b.append(4)
print(a)

# B
print(7 / 2, 7 // 2, 7 % 2)

# C
def adiciona(item, lista=[]):
    lista.append(item)
    return lista

print(adiciona("x"))
print(adiciona("y"))
```

````python
# A → [1, 2, 3, 4]   b e a são o mesmo objeto
a = [1, 2, 3]
b = a
b.append(4)
print(a)

# B → 3.5 3 1        divisão real, divisão inteira, resto
print(7 / 2, 7 // 2, 7 % 2)

# C → ['x'] e ['x', 'y']   o default mutável é criado UMA vez e fica guardado na função
def adiciona(item, lista=[]):
    lista.append(item)
    return lista

print(adiciona("x"))
print(adiciona("y"))
````

````python
# C corrigido
def adiciona(item, lista=None):
    if lista is None:
        lista = []
    lista.append(item)
    return lista


print(adiciona("x"))
print(adiciona("y"))
````

## Bônus

Refaça o exercício 7 usando o dataset `sns.load_dataset("tips")`: filtre as contas acima
de US$ 20 com uma comprehension sobre `df["total_bill"]`.

> A resolução mostra **as duas formas**: a comprehension que o enunciado pede e o filtro do
> pandas. A partir da aula 2, é sempre a segunda — percorrer uma coluna valor a valor é
> justamente o que o pandas existe para você não fazer.


````python
import seaborn as sns

df = sns.load_dataset("tips")

# mesma forma do exercício 7, só que a sequência agora é uma coluna do dataset:
# [ conta        for conta in df["total_bill"]   if conta > 20 ]
#   o que guardar    percorre a coluna valor a valor   só as contas acima de 20
altas = [conta for conta in df["total_bill"] if conta > 20]

print(len(altas))
print(altas[:10])

# ── e o jeito que a gente vai usar daqui para a frente ──────────────────
# o pandas filtra a coluna inteira de uma vez, sem laço nenhum:
altas_pandas = df[df["total_bill"] > 20]

print("\nmesmo resultado, sem percorrer nada:", len(altas_pandas))
````
