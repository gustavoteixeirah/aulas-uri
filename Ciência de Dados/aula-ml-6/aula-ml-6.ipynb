## ▸ Slides 8 a 10 · K-means

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

clientes = pd.read_csv("dados/Mall_Customers.csv")
colunas = ["Annual Income (k$)", "Spending Score (1-100)"]
X = StandardScaler().fit_transform(clientes[colunas])

inercias = []
for k in range(1, 11):
    ajuste = KMeans(n_clusters=k, n_init=10, random_state=0)
    ajuste.fit(X)
    inercias.append(ajuste.inertia_)
print([round(v) for v in inercias])

grupos = KMeans(n_clusters=5, n_init=10, random_state=0)
clientes["grupo"] = grupos.fit_predict(X)
centros = StandardScaler().fit(clientes[colunas]).inverse_transform(grupos.cluster_centers_)
for i, (renda, score) in enumerate(centros):
    n = int((clientes["grupo"] == i).sum())
    print(f"grupo {i}: renda {renda:.0f} · score {score:.0f} · {n} clientes")
```
